# 1. 리스트와 집합이란?

## 리스트 (List)
- **순서가 있는** 원소의 모음. 원소마다 **위치(인덱스)** 가 있다.
- **같은 값이 여러 번** 들어갈 수 있다.
- 질문: "`i`번째 값은?" → 위치로 바로 접근.

## 집합 (Set)
- **순서가 없는** 원소의 모음 (몇 번째라는 개념이 없다).
- **같은 값은 하나만** 존재한다.
- 질문: "이 값이 **들어 있나?**" → 소속 여부를 빠르게 판단.

| | 리스트 | 집합 |
|---|---|---|
| 비유 | 줄 선 사람들 (앞에서 몇 번째) | 동아리 명단 (들어 있나 없나) |
| 순서 | 있음 | 의미 없음 |
| 중복 | 허용 | 불가 |
| `x in 컨테이너` | 처음부터 훑는다 | 칸을 바로 찾아간다 |

> **멘토 노트**
> - 질문 1: "수강신청 대기 순번은 리스트? 집합?" → 순번이 중요하면 리스트, "신청했나 안 했나"만 보면 집합.
> - 질문 2: "방문한 도시 목록에서 **중복을 없애고 싶다**면?" → 집합. 단, 순서가 사라진다는 대가.
> - 흔한 오해: "set은 정렬되어 있다" ✗ — 작은 정수가 우연히 정렬돼 보일 뿐, 보장되지 않는다.

````python
L = [3, 1, 3, 2]
S = {3, 1, 3, 2}
print("리스트:", L, "길이", len(L))
print("집합  :", S, "길이", len(S))

vis.show(vis.array(L, name="리스트 L = [3, 1, 3, 2]"),
         vis.array(sorted(S), name="집합 S = {3, 1, 3, 2}  (순서 의미 없음)"),
         note="같은 재료인데 집합은 중복이 사라진다")

L.append(3); S.add(3)
vis.show(vis.array(L, name="L.append(3)", hl={len(L) - 1: "ok"}),
         vis.array(sorted(S), name="S.add(3)"),
         note="리스트는 5개로 늘고, 집합은 이미 있어서 그대로")
vis.inline(height=300)
````

````python
vis.show(vis.table(
    [["순서", "있음 (인덱스)", "없음"],
     ["중복", "허용", "불가"],
     ["i번째 접근  A[i]", "O(1)", "불가능"],
     ["x in A (검색)", "O(n)", "O(1) 평균"],
     ["맨 뒤 추가", "O(1) 평균", "O(1) 평균"],
     ["중간 삽입/삭제", "O(n)", "해당 없음"],
     ["원소 조건", "아무거나", "해시 가능한 값만"]],
    header=["", "리스트 list", "집합 set"], name="리스트 vs 집합"), note="오늘의 한 장 요약")
vis.inline(height=330)
````

````python
# "해시 가능한 값만" 이 무슨 뜻일까? — 직접 확인
S = {1, "a", (1, 2)}          # 숫자, 문자열, 튜플은 OK
print(S)
try:
    S.add([1, 2])              # 리스트는 바뀔 수 있어서(mutable) 해시를 못 구한다
except TypeError as e:
    print("TypeError:", e)
````

# 2. 리스트 표현 — 직접 코딩해 보기

파이썬의 `list`를 쓰기만 하지 말고, **속을 열어서** 같은 것을 만들어 봅니다.
같은 "리스트"라도 메모리에 놓는 방식이 두 가지입니다.

| 방식 | 생김새 | 장점 | 단점 |
|---|---|---|---|
| **배열 리스트** | 연속된 칸 | 인덱스 접근 O(1) | 중간 삽입·삭제 시 칸을 밀어야 함 |
| **연결 리스트** | 노드가 줄줄이 연결 | 앞쪽 삽입·삭제가 쉽다 | `i`번째로 가려면 처음부터 걸어가야 함 |

## 2-1. 배열 리스트 (ArrayList)
- 고정 크기 배열(`data`) + 실제 개수(`size`) + 칸 수(`capacity`)
- 꽉 차면 **2배 크기의 새 배열**로 옮겨 담는다 (동적 배열)
- 아래 `viz=True`는 시연용 옵션이고, 핵심 로직은 `viz`와 무관합니다.

````python
class ArrayList:
    def __init__(self, capacity=4, viz=False):
        self.capacity = capacity
        self.data = [None] * capacity      # 미리 잡아 둔 고정 크기 배열
        self.size = 0                      # 실제로 들어 있는 개수
        self.viz = viz

    def _v(self, note, hl=None):
        if self.viz:
            vis.show(vis.array(self.data, name=f"size={self.size}, capacity={self.capacity}",
                               hl=hl, ptr={"size": self.size}), note=note)

    def _grow(self):
        self.capacity *= 2
        new = [None] * self.capacity
        for i in range(self.size):
            new[i] = self.data[i]
        self.data = new
        self._v(f"꽉 참 → 용량을 {self.capacity}칸으로 2배 확장하고 옮겨 담음")

    def append(self, x):
        if self.size == self.capacity:
            self._grow()
        self.data[self.size] = x
        self.size += 1
        self._v(f"append({x}): 끝 칸에 쓰고 size + 1", {self.size - 1: "ok"})

    def insert(self, idx, x):
        if not 0 <= idx <= self.size:
            raise IndexError("범위를 벗어난 위치")
        if self.size == self.capacity:
            self._grow()
        for i in range(self.size, idx, -1):        # 뒤에서부터 한 칸씩 오른쪽으로 민다
            self.data[i] = self.data[i - 1]
            self._v(f"{i - 1}번 값을 {i}번으로 밀기", {i: "swp", i - 1: "cmp"})
        self.data[idx] = x
        self.size += 1
        self._v(f"insert({idx}, {x}): 빈 자리에 값을 쓴다", {idx: "ok"})

    def pop(self, idx=None):
        if self.size == 0:
            raise IndexError("빈 리스트")
        if idx is None:
            idx = self.size - 1
        if not 0 <= idx < self.size:
            raise IndexError("범위를 벗어난 위치")
        x = self.data[idx]
        for i in range(idx, self.size - 1):        # 빈자리를 메우려 한 칸씩 왼쪽으로 당긴다
            self.data[i] = self.data[i + 1]
            self._v(f"{i + 1}번 값을 {i}번으로 당기기", {i: "swp", i + 1: "cmp"})
        self.size -= 1
        self.data[self.size] = None
        self._v(f"pop({idx}) = {x}: 마지막 칸을 비움", {self.size: "dim"})
        return x

    def __getitem__(self, i):
        if not 0 <= i < self.size:
            raise IndexError("범위를 벗어난 위치")
        return self.data[i]

    def __len__(self):
        return self.size

    def __iter__(self):
        for i in range(self.size):
            yield self.data[i]

    def __repr__(self):
        return "ArrayList(" + repr(list(self)) + ")"
````

````python
a = ArrayList(capacity=4, viz=True)
for x in (10, 20, 30):
    a.append(x)
a.insert(1, 15)        # 중간 삽입 → 뒤쪽을 한 칸씩 밀어야 한다
a.append(40)           # 칸이 꽉 찼으니 확장!
a.pop(0)               # 맨 앞 삭제 → 전부 한 칸씩 당겨야 한다

assert list(a) == [15, 20, 30, 40] and len(a) == 4 and a[2] == 30
vis.inline(height=330)
````

> **멘토 노트 (ArrayList)**
> - 장면을 천천히 넘기며 묻기: "`insert(1, 15)`에서 왜 **뒤에서부터** 밀까?" → 앞에서부터 밀면 값이 덮어써진다.
> - "맨 앞 삽입/삭제가 느린 이유는?" → 모든 원소가 한 칸씩 움직여야 해서 O(n).
> - 파이썬 `list.append`가 "O(1) 평균"인 이유가 바로 아래 셀(용량 2배 확장)입니다.

````python
b = ArrayList(capacity=1)
caps = []
for i in range(17):
    b.append(i)
    caps.append(b.capacity)
print(caps)

vis.show(vis.bars(caps, name="append할 때마다의 capacity"),
         note="가끔 2배로 점프하지만, 대부분의 append는 칸만 채우면 끝 → 평균 O(1)")
vis.inline(height=300)
````

## 2-2. 연결 리스트 (Linked List)
- 값을 담은 **노드**(`data`)가 다음 노드를 가리키는 **링크**(`link`)로 이어진다.
- `head`(맨 앞 노드)만 기억하면 나머지는 링크를 따라 걸어갈 수 있다.
- 그림에서 노드는 `[값 | →]` 모양이고, 끝은 `None`입니다.

````python
class Node:
    def __init__(self, data, link=None):
        self.data = data
        self.link = link


class LinkedList:
    def __init__(self, viz=False):
        self.head = None
        self.viz = viz

    def _v(self, note, hl=None):
        if self.viz:
            vis.show(vis.linked(self.head, hl=hl), note=note)

    def push_front(self, x):
        self.head = Node(x, self.head)                  # 새 노드가 기존 head를 가리키게
        self._v(f"push_front({x}): 새 노드를 맨 앞에 달고 head 갱신", {0: "ok"})

    def append(self, x):
        node = Node(x)
        if self.head is None:
            self.head = node
            self._v(f"append({x}): 빈 리스트라 head가 된다", {0: "ok"})
            return
        cur, pos = self.head, 0
        while cur.link is not None:                     # 끝 노드를 찾을 때까지 걸어간다
            self._v(f"끝을 찾는 중 ({pos}번째 노드)", {pos: "cmp"})
            cur, pos = cur.link, pos + 1
        cur.link = node
        self._v(f"append({x}): 끝 노드의 link를 새 노드로", {pos + 1: "ok"})

    def find(self, x):
        cur, pos = self.head, 0
        while cur is not None:
            if cur.data == x:
                self._v(f"{pos}번째에서 {x} 발견!", {pos: "ok"})
                return pos
            self._v(f"{pos}번째 값 {cur.data} ≠ {x}", {pos: "cmp"})
            cur, pos = cur.link, pos + 1
        self._v(f"{x}는 없다 (끝까지 가 봄)")
        return -1

    def remove(self, x):
        prev, cur, pos = None, self.head, 0
        while cur is not None:
            if cur.data == x:
                if prev is None:
                    self.head = cur.link                # 맨 앞 삭제: head만 옮기면 끝
                else:
                    prev.link = cur.link                # 앞 노드가 뒤 노드를 직접 가리키게
                self._v(f"remove({x}): 앞 노드의 link를 건너뛰게 바꿈", {pos - 1: "ok"} if prev else {0: "ok"})
                return True
            prev, cur, pos = cur, cur.link, pos + 1
        return False

    def __iter__(self):
        cur = self.head
        while cur is not None:
            yield cur.data
            cur = cur.link

    def __repr__(self):
        return "LinkedList(" + repr(list(self)) + ")"
````

````python
ll = LinkedList(viz=True)
for x in (10, 20, 30):
    ll.append(x)
ll.push_front(5)       # 앞쪽 삽입은 칸을 밀 필요 없이 O(1)
ll.find(20)            # 찾으려면 처음부터 걸어가야 한다
ll.remove(10)          # 링크만 바꾸면 삭제 끝 (위치를 알고 있다면 O(1))

assert list(ll) == [5, 20, 30]
vis.inline(height=300)
````

## 2-3. 집합 표현 — 해시 테이블 (체이닝)
집합이 `x in S`를 빠르게 하는 비결은 **값으로 칸 번호를 계산**하는 것입니다.

1. `칸 번호 = hash(x) % 버킷 수`
2. 그 칸(버킷)만 들여다보면 된다 → 전부 훑지 않아도 됨
3. 서로 다른 값이 같은 칸에 몰리면(**충돌**) 그 칸 안에 줄 세워 둔다 (체이닝)

````python
class MySet:
    def __init__(self, nbuckets=7, viz=False):
        self.buckets = [[] for _ in range(nbuckets)]
        self.size = 0
        self.viz = viz

    def _v(self, note, hl=None):
        if self.viz:
            vis.show(vis.hashtable(self.buckets, name=f"size={self.size}", hl=hl), note=note)

    def _idx(self, x):
        return hash(x) % len(self.buckets)

    def add(self, x):
        i = self._idx(x)
        if x in self.buckets[i]:
            self._v(f"add({x}): {i}번 버킷에 이미 있음 → 무시 (집합은 중복 불가)", {i: "dim"})
            return
        self.buckets[i].append(x)
        self.size += 1
        self._v(f"add({x}): {x} % {len(self.buckets)} = {i}번 버킷에 넣음", {i: "ok"})

    def remove(self, x):
        i = self._idx(x)
        if x not in self.buckets[i]:
            raise KeyError(x)
        self.buckets[i].remove(x)
        self.size -= 1
        self._v(f"remove({x}): {i}번 버킷에서 뺌", {i: "swp"})

    def __contains__(self, x):
        return x in self.buckets[self._idx(x)]       # 한 버킷만 본다

    def __len__(self):
        return self.size

    def __iter__(self):
        for b in self.buckets:
            yield from b
````

````python
s = MySet(nbuckets=7, viz=True)
for x in (10, 3, 17, 24, 10, 7):       # 10, 3, 17, 24는 모두 7로 나눈 나머지가 3 → 충돌!
    s.add(x)
s.remove(17)

assert len(s) == 4 and 3 in s and 17 not in s and 5 not in s
print(sorted(s))
vis.inline(height=420)
````

> **멘토 노트 (MySet)**
> - "10, 3, 17, 24가 전부 3번 칸에 몰렸다. 이러면 `in`이 빠를까?" → 그 칸 안에서는 훑어야 하므로 느려진다.
>   그래서 실제 `set`은 칸 수를 넉넉히 유지하고(리해시) 값이 고르게 퍼지도록 해시 함수를 설계한다 → "평균 O(1)".
> - 정수는 `hash(x) == x`라 재현이 쉽지만, 문자열 해시는 실행할 때마다 달라질 수 있어 데모에는 정수를 썼습니다.
> - 이 구현에는 **리해시(칸 늘리기)** 가 없습니다. 심화 과제로 내기 좋습니다.

## 3-1. 리스트 슬라이싱 — `A[시작:끝:간격]`

````python
A = [10, 20, 30, 40, 50, 60]
print("A[1:4]  =", A[1:4])
print("A[-2:]  =", A[-2:])
print("A[::2]  =", A[::2])
print("A[::-1] =", A[::-1])

vis.show(vis.array(A, name="A"), note="A = [10, 20, 30, 40, 50, 60]")
vis.show(vis.array(A, name="A", hl={1: "cmp", 2: "cmp", 3: "cmp"}), note="A[1:4] → 1·2·3번 (4번은 포함하지 않음)")
vis.show(vis.array(A, name="A", hl={4: "cur", 5: "cur"}), note="A[-2:] → 뒤에서 2개")
vis.show(vis.array(A, name="A", hl={i: "piv" for i in range(0, 6, 2)}), note="A[::2] → 한 칸 건너뛰며")
vis.inline(height=260)
````

````python
# 자주 쓰는 리스트 연산 모음
nums = [4, 1, 3, 1, 5]

nums.append(9)              # 뒤에 추가
nums.insert(0, 7)           # 맨 앞 삽입 (O(n))
last = nums.pop()           # 맨 뒤 꺼내기 (O(1))
nums.remove(1)              # 값으로 첫 번째 하나 삭제
print(nums, last)

print(sorted(nums), nums)               # sorted: 새 리스트 / sort(): 제자리 정렬
squares = [x * x for x in range(1, 6)]  # 리스트 컴프리헨션
evens = [x for x in range(10) if x % 2 == 0]
print(squares, evens)

for i, v in enumerate(["a", "b", "c"]):
    print(i, v, end=" | ")
print()
print(nums.count(1), nums.index(5), len(nums), sum(nums), max(nums), min(nums))
````

## 3-2. 2차원 리스트의 함정
`[[0] * 3] * 3`은 **같은 행을 3번 가리키는** 리스트입니다. 한 칸을 바꾸면 모든 행이 바뀝니다.

````python
bad = [[0] * 3] * 3
good = [[0] * 3 for _ in range(3)]
bad[0][0] = 1
good[0][0] = 1
print("bad :", bad)
print("good:", good)
print("bad의 세 행은 같은 객체?", bad[0] is bad[1] is bad[2])

vis.show(vis.table(bad, name="bad = [[0]*3]*3", hl={0: "swp", 1: "swp", 2: "swp"}),
         vis.table(good, name="good = [[0]*3 for _ in range(3)]", hl={0: "ok"}),
         note="[0][0]만 1로 바꿨는데, bad는 모든 행이 바뀐다")
vis.inline(height=260)
````

> **멘토 노트**
> - 코딩테스트(격자, DP 표)에서 정말 자주 터지는 실수입니다. "왜 이렇게 되는지"를 **같은 객체를 가리킨다**로 설명.
> - 슬라이싱 `B = A[:]`은 **얕은 복사**. 안에 리스트가 있으면 안쪽은 여전히 공유됩니다 (`copy.deepcopy`).

## 3-3. 집합 연산 — 합집합 · 교집합 · 차집합 · 대칭차

````python
A = {1, 2, 3, 4}
B = {3, 4, 5}
ops = [("A | B  합집합", A | B), ("A & B  교집합", A & B),
       ("A - B  차집합", A - B), ("A ^ B  대칭차집합 (한쪽에만 있는 것)", A ^ B)]
for title, res in ops:
    print(title, "=", sorted(res))
    vis.show(vis.array(sorted(A), name="A"), vis.array(sorted(B), name="B"),
             vis.array(sorted(res), name=title, done=range(len(res))), note=title)
vis.inline(height=330)
````

## 3-4. 왜 집합을 쓰나? — `in` 속도 비교

````python
import time

N = 50_000
big_list = list(range(N))
big_set = set(big_list)
targets = [N - 1] * 200            # 리스트에서 가장 오래 걸리는 값(맨 끝)을 200번 찾기

t0 = time.perf_counter()
for t in targets:
    t in big_list
t_list = time.perf_counter() - t0

t0 = time.perf_counter()
for t in targets:
    t in big_set
t_set = time.perf_counter() - t0

print(f"list : {t_list * 1000:8.2f} ms")
print(f"set  : {t_set * 1000:8.2f} ms   (약 {t_list / max(t_set, 1e-9):,.0f}배 차이)")

vis.show(vis.bars([t_list * 1000, t_set * 1000], name="x in 컨테이너 200번 (ms) — 왼쪽: list, 오른쪽: set",
                  hl={0: "swp", 1: "ok"}), note="값이 많을수록 차이는 더 벌어진다")
vis.inline(height=300)
````

## 3-5. 중복 제거 — 순서를 지키고 싶다면?

````python
words = ["b", "a", "b", "c", "a", "d"]

print(set(words))                       # 순서는 보장되지 않는다
print(list(dict.fromkeys(words)))       # 처음 나온 순서를 유지하며 중복 제거 (3.7+)

seen, uniq = set(), []                  # 직접 쓰면: 집합으로 "봤나?"만 빠르게 확인
for w in words:
    if w not in seen:
        seen.add(w)
        uniq.append(w)
    vis.show(vis.array(words, hl={words.index(w): "cur"}), vis.array(uniq, name="uniq (순서 유지)"),
             vis.array(sorted(seen), name="seen (봤나 확인용 집합)"), note=f"'{w}' 확인")
vis.inline(height=380)
````

## 3-6. 리스트로 정렬하기 — 선택 정렬 시각화
막대그래프로 "가장 작은 값을 찾아 앞으로 보내는" 과정을 봅니다. (보라 = 현재 최솟값, 주황 = 비교 중, 초록 = 확정)

````python
A = [7, 3, 9, 1, 6, 2, 8, 4]
vis.show(vis.bars(A), note="처음 상태")
for i in range(len(A) - 1):
    m = i
    for j in range(i + 1, len(A)):
        vis.show(vis.bars(A, hl={m: "piv", j: "cmp"}, done=range(i)),
                 note=f"{i}번 자리에 올 최솟값 찾는 중: A[{j}]={A[j]} vs 현재 최솟값 A[{m}]={A[m]}")
        if A[j] < A[m]:
            m = j
    A[i], A[m] = A[m], A[i]
    vis.show(vis.bars(A, hl={i: "ok"}, done=range(i)), note=f"A[{i}] ↔ A[{m}] 교환")
vis.show(vis.bars(A, done=range(len(A))), note="정렬 끝")
print(A)
vis.inline(height=360)
````

## 문제 1. 백준 10818 — 최소, 최대

**문제 요약**: 정수 `N`개가 주어진다. 이 중 **최솟값과 최댓값**을 출력한다.

```
입력                      출력
5                         7 35
20 10 35 30 7
```

**접근**: 첫 값을 최소/최대로 놓고, 나머지를 한 번씩 훑으며 갱신한다. (O(N))

````python
def min_max(nums):
    mn = mx = nums[0]
    for x in nums[1:]:
        if x < mn:
            mn = x
        if x > mx:
            mx = x
    return mn, mx


assert min_max([20, 10, 35, 30, 7]) == (7, 35)
assert min_max([5]) == (5, 5)
assert min_max([-3, -1, -9]) == (-9, -1)

# ---- vis로 따라가기 (같은 로직에 장면만 끼워 넣음)
nums = [20, 10, 35, 30, 7]
mn = mx = nums[0]
vis.show(vis.array(nums, hl={0: "cur"}), vis.vars(min=mn, max=mx), note="첫 값으로 시작")
for i in range(1, len(nums)):
    x = nums[i]
    if x < mn:
        mn = x
    if x > mx:
        mx = x
    vis.show(vis.array(nums, hl={i: "cur"}), vis.vars(min=mn, max=mx), note=f"{x} 확인")
vis.inline(height=290)
````

입력                              출력

**접근**: 나머지를 **집합**에 넣으면 중복이 알아서 사라진다. 마지막에 `len`.

````python
def count_distinct_mod(nums, k=42):
    return len({x % k for x in nums})


assert count_distinct_mod([39, 40, 41, 42, 43, 44, 82, 83, 84, 85]) == 6
assert count_distinct_mod(list(range(1, 11))) == 10
assert count_distinct_mod([42, 84, 126, 168, 210, 252, 294, 336, 378, 420]) == 1

nums = [39, 40, 41, 42, 43, 44, 82, 83, 84, 85]
seen = set()
for x in nums:
    r = x % 42
    dup = r in seen
    seen.add(r)
    vis.show(vis.vars(수=x, 나머지=r), vis.array(sorted(seen), name=f"나머지 집합 (현재 {len(seen)}개)"),
             note=f"{x} % 42 = {r}" + (" → 이미 있어서 개수 그대로" if dup else " → 새 값!"))
vis.inline(height=290)
````

## 문제 4. 백준 1181 — 단어 정렬 (리스트 + 집합 + 정렬)

**문제 요약**: 알파벳 소문자 단어 `N`개가 주어진다. **중복을 제거**하고, ① **길이가 짧은 것부터**, ② 길이가 같으면 **사전 순**으로 출력한다.

```
입력 예 (단어들)                                                    출력
but i wont hesitate no more no more hesitate cannot wont           i / no / but / more / wont / cannot / hesitate
```

**접근**: `set`으로 중복 제거 → `sorted(..., key=lambda w: (len(w), w))`. 키가 **튜플**이면 앞 원소부터 비교한다.

````python
def sort_words(words):
    return sorted(set(words), key=lambda w: (len(w), w))


words = ["but", "i", "wont", "hesitate", "no", "more", "no", "more", "hesitate", "cannot", "wont"]
assert sort_words(words) == ["i", "no", "but", "more", "wont", "cannot", "hesitate"]
assert sort_words(["b", "a", "a"]) == ["a", "b"]

step1 = list(words)
step2 = list(set(words))
step3 = sort_words(words)
vis.show(vis.array(step1, name="① 입력 (중복 있음)"), note=f"{len(step1)}개")
vis.show(vis.array(step1, name="① 입력"), vis.array(sorted(set(words)), name="② set으로 중복 제거 (사전순으로 늘어놓음)"),
         note=f"{len(set(words))}개로 줄었다")
vis.show(vis.array(step3, name="③ (길이, 사전순)으로 정렬", done=range(len(step3))),
         vis.table([[w, len(w)] for w in step3], header=["단어", "길이"]), note="길이 → 사전순")
vis.inline(height=420)
````

## 문제 5. 프로그래머스 — 같은 숫자는 싫어 (Lv.1)

**문제 요약**: 0~9 숫자로 된 배열 `arr`에서 **연속으로 나타나는 같은 숫자는 하나만** 남긴다. 원래 순서는 유지한다.

```
arr                     return
[1, 1, 3, 3, 0, 1, 1]   [1, 3, 0, 1]
[4, 4, 4, 3, 3]         [4, 3]
```

**접근**: 결과 리스트의 **마지막 값**과 지금 값을 비교해, 다를 때만 `append`. (집합을 쓰면 `[1,3,0,1]`의 두 번째 1이 사라지므로 **안 된다** — 연속만 지우는 문제)

````python
def solution(arr):
    answer = []
    for x in arr:
        if not answer or answer[-1] != x:
            answer.append(x)
    return answer


assert solution([1, 1, 3, 3, 0, 1, 1]) == [1, 3, 0, 1]
assert solution([4, 4, 4, 3, 3]) == [4, 3]
assert solution([7]) == [7]

arr = [1, 1, 3, 3, 0, 1, 1]
answer = []
for i, x in enumerate(arr):
    keep = not answer or answer[-1] != x
    vis.show(vis.array(arr, hl={i: "cur"}), vis.array(answer, name="answer"),
             note=f"{x} vs answer의 마지막({answer[-1] if answer else '없음'}) → " + ("다르니 추가" if keep else "같으니 건너뜀"))
    if keep:
        answer.append(x)
vis.show(vis.array(arr), vis.array(answer, name="answer", done=range(len(answer))), note="완성")
vis.inline(height=330)
````

> **멘토 노트**: "왜 `set`으로 안 될까?"를 꼭 질문하세요. **조건을 정확히 읽는 습관**(연속인가, 전체인가)을 만드는 문제입니다.

## 문제 6. 프로그래머스 — 폰켓몬 (Lv.1, 집합)

**문제 요약**: 폰켓몬 `N`마리(`N`은 짝수)가 번호로 주어진다(같은 번호 = 같은 종류). 이 중 **`N/2`마리**를 고를 때, 가질 수 있는 **종류 수의 최댓값**을 구한다.

```
nums                  result
[3, 1, 2, 3]          2
[3, 3, 3, 2, 2, 2]    2
[3, 3, 3, 2, 2, 4]    3
```

**접근**: 종류 수 = `len(set(nums))`, 고를 수 있는 마리 수 = `N // 2`. 둘 중 **작은 쪽**이 답이다.

````python
def solution(nums):
    return min(len(set(nums)), len(nums) // 2)


assert solution([3, 1, 2, 3]) == 2
assert solution([3, 3, 3, 2, 2, 2]) == 2
assert solution([3, 3, 3, 2, 2, 4]) == 3

for nums in ([3, 1, 2, 3], [3, 3, 3, 2, 2, 2], [3, 3, 3, 2, 2, 4]):
    kinds, pick = len(set(nums)), len(nums) // 2
    vis.show(vis.array(nums, name="nums"), vis.array(sorted(set(nums)), name="종류 (set)"),
             vis.vars(종류수=kinds, 고를수_N절반=pick, 답=min(kinds, pick)),
             note=f"종류 {kinds}, 고를 수 있는 건 {pick}마리 → 답 {min(kinds, pick)}")
vis.inline(height=330)
````

> **멘토 노트**: "종류가 5개인데 3마리만 고를 수 있다면? 종류가 2개인데 3마리 고를 수 있다면?" 두 경우로 나눠 생각하게 하면 `min`이 자연스럽게 나옵니다.

## 문제 7. 프로그래머스 — 두 개 뽑아서 더하기 (Lv.1, 리스트 + 집합)

**문제 요약**: 정수 배열 `numbers`에서 **서로 다른 인덱스**의 두 수를 뽑아 더해서 만들 수 있는 모든 수를 **중복 없이 오름차순**으로 반환한다.

```
numbers          result
[2, 1, 3, 4, 1]  [2, 3, 4, 5, 6, 7]
[5, 0, 2, 7]     [2, 5, 7, 9, 12]
```

**접근**: 두 인덱스 `(i, j)` (`i < j`)를 모두 훑어 합을 **집합**에 넣고, 마지막에 정렬한다.

````python
def solution(numbers):
    sums = set()
    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            sums.add(numbers[i] + numbers[j])
    return sorted(sums)


assert solution([2, 1, 3, 4, 1]) == [2, 3, 4, 5, 6, 7]
assert solution([5, 0, 2, 7]) == [2, 5, 7, 9, 12]

numbers = [2, 1, 3, 4, 1]
sums = set()
for i in range(len(numbers)):
    for j in range(i + 1, len(numbers)):
        s = numbers[i] + numbers[j]
        new = s not in sums
        sums.add(s)
        vis.show(vis.array(numbers, hl={i: "cmp", j: "cmp"}), vis.array(sorted(sums), name="합의 집합"),
                 note=f"numbers[{i}] + numbers[{j}] = {s}" + (" (새 값)" if new else " (이미 있음)"))
vis.show(vis.array(sorted(sums), name="정렬해서 반환", done=range(len(sums))), note="완성")
vis.inline(height=300)
````

> **멘토 노트**: `i < j`로 `j`를 `i + 1`부터 시작하는 이유(같은 쌍 중복·자기 자신 방지)를 짚어 주세요.
